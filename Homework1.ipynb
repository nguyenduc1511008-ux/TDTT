```python
#8

def tinh_tien_dien():
    ho_ten = input("Ten chu ho: ")
    cs_truoc = int(input("Chi so thang truoc: "))
    cs_nay = int(input("Chi so thang nay: "))
    
    kwh = cs_nay - cs_truoc
    
    tien_chua_vat = 0
    
    if kwh <= 50:
        tien_chua_vat = kwh * 1984
    elif kwh <= 100:
        tien_chua_vat = (50 * 1984) + (kwh - 50) * 2050
    elif kwh <= 200:
        tien_chua_vat = (50 * 1984) + (50 * 2050) + (kwh - 100) * 2380
    elif kwh <= 300:
        tien_chua_vat = (50 * 1984) + (50 * 2050) + (100 * 2380) + (kwh - 200) * 2998
    elif kwh <= 400:
        tien_chua_vat = (50 * 1984) + (50 * 2050) + (100 * 2380) + (100 * 2998) + (kwh - 300) * 3350
    else:
        tien_chua_vat = (50 * 1984) + (50 * 2050) + (100 * 2380) + (100 * 2998) + (100 * 3350) + (kwh - 400) * 3460
    tong_tien = tien_chua_vat * 1.08
    print(f"Ho va ten: {ho_ten}")
    print(f"Tien phai tra la: {tong_tien:.0f}")
tinh_tien_dien()
```
